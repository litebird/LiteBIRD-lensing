#### Show inhonogenity of noise after component separation

```python
import numpy as np, healpy as hp, cmb, local, warnings
warnings.filterwarnings("ignore")
```

```python
# define parameters
NSIDE = 64
pobj = local.analysis()
simn = 20
```

```python
QU = np.array( [ hp.ud_grade( hp.read_map(pobj.ffgs[rlz],field=(1,2)), NSIDE ) for rlz in range(1,simn+1) ] ) / cmb.Tcmb
mQU = np.mean(QU,axis=0)
sQU = np.std(QU,axis=0)
```

```python
M = hp.ud_grade(hp.read_map('../data/lensing/FG_mask.fits'),NSIDE)
```

```python
hp.mollview(M*mQU[0],coord=["G","C"],min=-3e-8,max=3e-8)
```

```python
hp.mollview(M*sQU[0],coord=["G","C"],min=0.,max=3e-8)
```

```python
hp.cartview(M*sQU[0],min=0,max=3e-8,coord=["G","C"],lonra=[-70,70],latra=[-80,-20])
```
